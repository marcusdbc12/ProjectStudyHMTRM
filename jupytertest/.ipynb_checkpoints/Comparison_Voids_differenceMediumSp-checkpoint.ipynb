```python
# ============================================================
# CELL 1 — LIBRARIES
# ============================================================

import os
import glob
import re

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
# ============================================================
# CELL 2 — OUTPUT FOLDER
# ============================================================

comparison_dir = "cmp_limit"

os.makedirs(comparison_dir, exist_ok=True)

fig_dir = os.path.join(comparison_dir, "figs")
os.makedirs(fig_dir, exist_ok=True)

print("Comparison output folder:", comparison_dir)
print("Comparison figure folder:", fig_dir)


def save_current_figure(figure_name, dpi=300):
    safe_name = (
        figure_name
        .lower()
        .replace(" ", "_")
        .replace(":", "")
        .replace("→", "to")
        .replace("/", "_")
        .replace("\\", "_")
    )

    save_path = os.path.join(fig_dir, f"{safe_name}.png")

    plt.savefig(
        save_path,
        dpi=dpi,
        bbox_inches="tight"
    )

    print("Saved figure:", save_path)

    return save_path
```

```python
# ============================================================
# CELL 3 — FIND NEW CONTRAST-LIMIT FOLDERS
# ============================================================

si_speed = 8433.0

benchmark_folders = sorted(
    glob.glob("fwi_void_20to85_v*_c*_u*")
)

print("Found folders:")
for folder in benchmark_folders:
    print(folder)

print("Number of folders:", len(benchmark_folders))
```

```python
# ============================================================
# CELL 4 — EXTRACT METADATA FROM FOLDER NAME
# ============================================================

def extract_folder_value(folder_name, pattern):
    match = re.search(pattern, folder_name)
    if match:
        return float(match.group(1))
    return np.nan


def extract_void_speed_percent(folder_name):
    # Example: fwi_void_20to85_v90_c10_u843 -> 90
    return extract_folder_value(folder_name, r"_v(\d+)")


def extract_contrast_percent(folder_name):
    # Example: fwi_void_20to85_v90_c10_u843 -> 10
    return extract_folder_value(folder_name, r"_c(\d+)")


def extract_max_update(folder_name):
    # Example: fwi_void_20to85_v90_c10_u843 -> 843
    return extract_folder_value(folder_name, r"_u(\d+)")


def add_folder_metadata(df, folder):
    void_speed_percent = extract_void_speed_percent(folder)
    contrast_percent = extract_contrast_percent(folder)
    max_update = extract_max_update(folder)

    df["benchmark_folder"] = folder

    df["folder_void_speed_percent_of_silicon"] = void_speed_percent
    df["folder_void_speed_fraction"] = void_speed_percent / 100.0

    df["folder_contrast_percent_relative_to_silicon"] = contrast_percent
    df["folder_contrast_fraction_relative_to_silicon"] = contrast_percent / 100.0

    df["folder_max_velocity_update_m_per_s"] = max_update
    df["folder_max_update_percent_of_silicon"] = 100.0 * max_update / si_speed

    df["folder_true_void_speed_m_per_s"] = si_speed * void_speed_percent / 100.0
    df["folder_speed_difference_from_silicon_m_per_s"] = si_speed * contrast_percent / 100.0

    return df
```

```python
# ============================================================
# CELL 5 — READ ALL RESULT CSV FILES
# ============================================================

all_metadata = []
all_void_metrics = []
all_global_metrics = []
all_shape_metrics = []
all_evidence = []

for folder in benchmark_folders:

    metadata_csv = os.path.join(
        folder,
        "TABLE_benchmark_metadata_contrast_limit.csv"
    )

    void_csv = os.path.join(
        folder,
        "TABLE_circular_void_scientific_metrics_READABLE.csv"
    )

    global_csv = os.path.join(
        folder,
        "TABLE_global_reconstruction_metrics_READABLE.csv"
    )

    shape_csv = os.path.join(
        folder,
        "TABLE_circular_void_shape_boundary_deviation_READABLE.csv"
    )

    evidence_csv = os.path.join(
        folder,
        "TABLE_final_void_evidence_summary_READABLE.csv"
    )

    if os.path.exists(metadata_csv):
        df = pd.read_csv(metadata_csv)
        df = add_folder_metadata(df, folder)
        all_metadata.append(df)
    else:
        print("Missing metadata:", metadata_csv)

    if os.path.exists(void_csv):
        df = pd.read_csv(void_csv)
        df = add_folder_metadata(df, folder)
        all_void_metrics.append(df)
    else:
        print("Missing void metrics:", void_csv)

    if os.path.exists(global_csv):
        df = pd.read_csv(global_csv)
        df = add_folder_metadata(df, folder)
        all_global_metrics.append(df)
    else:
        print("Missing global metrics:", global_csv)

    if os.path.exists(shape_csv):
        df = pd.read_csv(shape_csv)
        df = add_folder_metadata(df, folder)
        all_shape_metrics.append(df)
    else:
        print("Missing shape metrics:", shape_csv)

    if os.path.exists(evidence_csv):
        df = pd.read_csv(evidence_csv)
        df = add_folder_metadata(df, folder)
        all_evidence.append(df)
    else:
        print("Missing evidence summary:", evidence_csv)


metadata_all = pd.concat(all_metadata, ignore_index=True)
void_metrics_all = pd.concat(all_void_metrics, ignore_index=True)
global_metrics_all = pd.concat(all_global_metrics, ignore_index=True)
shape_metrics_all = pd.concat(all_shape_metrics, ignore_index=True)

if len(all_evidence) > 0:
    evidence_all = pd.concat(all_evidence, ignore_index=True)
else:
    evidence_all = pd.DataFrame()

print("metadata_all:", metadata_all.shape)
print("void_metrics_all:", void_metrics_all.shape)
print("global_metrics_all:", global_metrics_all.shape)
print("shape_metrics_all:", shape_metrics_all.shape)
print("evidence_all:", evidence_all.shape)

display(void_metrics_all.head())
display(shape_metrics_all.head())
```

```python
# ============================================================
# CELL 6 — STAGE 2 CONTRAST-LIMIT SUMMARY
# ============================================================

stage2_void = void_metrics_all[
    void_metrics_all["FWI stage"] == "Stage 2"
].copy()

stage2_shape = shape_metrics_all[
    shape_metrics_all["FWI stage"] == "Stage 2"
].copy()

stage2_void_keep = [
    "benchmark_folder",

    "folder_void_speed_percent_of_silicon",
    "folder_void_speed_fraction",
    "folder_contrast_percent_relative_to_silicon",
    "folder_contrast_fraction_relative_to_silicon",
    "folder_true_void_speed_m_per_s",
    "folder_speed_difference_from_silicon_m_per_s",
    "folder_max_velocity_update_m_per_s",

    "void_speed_fraction",
    "true_void_speed_m_per_s",
    "required_update_m_per_s",

    "reconstructed_mean_void_speed_m_per_s",
    "reconstructed_mean_void_update_m_per_s",
    "void_velocity_update_recovery_percent",

    "void_MAE_m_per_s",
    "void_RMSE_m_per_s",
    "background_MAE_m_per_s",
    "background_RMSE_m_per_s",

    "void_detection_efficiency_0_to_1",
    "false_positive_ratio_background_to_void",
    "wrong_sign_ratio_inside_void",

    "recovery_ok",
    "localization_ok",
    "false_positive_ok",
    "wrong_sign_ok",
    "background_ok",
    "successful_detection",
]

stage2_shape_keep = [
    "benchmark_folder",

    "IoU_intersection_over_union",
    "precision_detected_pixels_inside_true_void",
    "recall_true_void_pixels_detected",
    "F1_score",

    "center_error_micrometers",
    "radius_error_micrometers",

    "detected_void_area_pixels",
    "true_void_area_pixels",
]

stage2_summary = pd.merge(
    stage2_void[stage2_void_keep],
    stage2_shape[stage2_shape_keep],
    on="benchmark_folder",
    how="left"
)

stage2_summary = stage2_summary.sort_values(
    "folder_contrast_percent_relative_to_silicon"
)

display(stage2_summary)

summary_csv = os.path.join(
    comparison_dir,
    "summary_stage2_contrast_limit.csv"
)

stage2_summary.to_csv(summary_csv, index=False)

print("Saved:")
print(summary_csv)
```

```python
# ============================================================
# CELL 7 — SCIENTIFIC PASS / FAIL CRITERIA
# ============================================================

# These are adjustable scientific thresholds.


recovery_min_percent = 70.0
recovery_max_percent = 130.0

min_detection_efficiency = 0.60
max_false_positive_ratio = 0.50
max_wrong_sign_ratio = 0.30

min_iou = 0.25
min_f1_score = 0.25

max_background_mae_fraction = 0.25


range_tolerance_m_per_s = 1.0

stage2_summary["range_sufficient"] = (
    stage2_summary["folder_max_velocity_update_m_per_s"]
    >= stage2_summary["required_update_m_per_s"] - range_tolerance_m_per_s

)

stage2_summary["recovery_pass"] = (
    (stage2_summary["void_velocity_update_recovery_percent"] >= recovery_min_percent)
    & (stage2_summary["void_velocity_update_recovery_percent"] <= recovery_max_percent)
)

stage2_summary["localization_pass"] = (
    stage2_summary["void_detection_efficiency_0_to_1"]
    >= min_detection_efficiency
)

stage2_summary["false_positive_pass"] = (
    stage2_summary["false_positive_ratio_background_to_void"]
    <= max_false_positive_ratio
)

stage2_summary["wrong_sign_pass"] = (
    stage2_summary["wrong_sign_ratio_inside_void"]
    <= max_wrong_sign_ratio
)

stage2_summary["shape_pass"] = (
    (stage2_summary["IoU_intersection_over_union"] >= min_iou)
    & (stage2_summary["F1_score"] >= min_f1_score)
)

stage2_summary["background_pass"] = (
    stage2_summary["background_MAE_m_per_s"]
    <= max_background_mae_fraction * stage2_summary["required_update_m_per_s"]
)

stage2_summary["overall_quality_pass"] = (
    stage2_summary["range_sufficient"]
    & stage2_summary["recovery_pass"]
    & stage2_summary["localization_pass"]
    & stage2_summary["false_positive_pass"]
    & stage2_summary["wrong_sign_pass"]
    & stage2_summary["shape_pass"]
    & stage2_summary["background_pass"]
)


def classify_failure(row):
    if row["overall_quality_pass"]:
        return "pass"

    if not row["range_sufficient"]:
        return "range_limited_or_rounding_issue"

    if not row["recovery_pass"]:
        if row["void_velocity_update_recovery_percent"] < 0:
            return "wrong_direction_amplitude_failure"
        else:
            return "amplitude_recovery_failure"

    if not row["wrong_sign_pass"]:
        return "wrong_sign_failure"

    if not row["background_pass"]:
        return "background_artifact_failure"

    if not row["shape_pass"]:
        return "shape_localization_failure"

    if not row["localization_pass"]:
        return "weak_void_localization"

    if not row["false_positive_pass"]:
        return "false_positive_failure"

    return "quality_failure"


stage2_summary["quality_class"] = stage2_summary.apply(
    classify_failure,
    axis=1
)

display(stage2_summary[
    [
        "folder_contrast_percent_relative_to_silicon",
        "folder_void_speed_percent_of_silicon",
        "folder_true_void_speed_m_per_s",
        "folder_speed_difference_from_silicon_m_per_s",

        "void_velocity_update_recovery_percent",
        "void_MAE_m_per_s",
        "background_MAE_m_per_s",
        "void_detection_efficiency_0_to_1",
        "false_positive_ratio_background_to_void",
        "wrong_sign_ratio_inside_void",
        "IoU_intersection_over_union",
        "F1_score",

        "overall_quality_pass",
        "quality_class",
    ]
])

quality_csv = os.path.join(
    comparison_dir,
    "summary_stage2_quality_pass_fail.csv"
)

stage2_summary.to_csv(quality_csv, index=False)

print("Saved:")
print(quality_csv)
```

```python
# ============================================================
# CELL 8 — FIND MAXIMUM SUPPORTED SPEED DIFFERENCE
# ============================================================

passed_cases = stage2_summary[
    stage2_summary["overall_quality_pass"] == True
].copy()

failed_cases = stage2_summary[
    stage2_summary["overall_quality_pass"] == False
].copy()

if len(passed_cases) > 0:
    max_supported_contrast_percent = passed_cases[
        "folder_contrast_percent_relative_to_silicon"
    ].max()

    max_supported_speed_difference = passed_cases[
        "folder_speed_difference_from_silicon_m_per_s"
    ].max()

    max_supported_row = passed_cases.sort_values(
        "folder_contrast_percent_relative_to_silicon"
    ).iloc[-1]

    print("MAXIMUM SUPPORTED VOID-SPEED CONTRAST:")
    print(f"{max_supported_contrast_percent:.2f}% relative to silicon")

    print("\nMAXIMUM SUPPORTED SPEED DIFFERENCE:")
    print(f"{max_supported_speed_difference:.2f} m/s")

    print("\nCorresponding void speed:")
    print(f"{max_supported_row['folder_true_void_speed_m_per_s']:.2f} m/s")

    print("\nBest supported folder:")
    print(max_supported_row["benchmark_folder"])

else:
    max_supported_contrast_percent = np.nan
    max_supported_speed_difference = np.nan

    print("No contrast passed the selected quality criteria.")

# ------------------------------------------------------------
# First failure after the last pass
# ------------------------------------------------------------

if len(passed_cases) > 0:
    first_failure_after_limit = failed_cases[
        failed_cases["folder_contrast_percent_relative_to_silicon"]
        > max_supported_contrast_percent
    ]

    if len(first_failure_after_limit) > 0:
        first_failure_after_limit = first_failure_after_limit.sort_values(
            "folder_contrast_percent_relative_to_silicon"
        ).iloc[0]

        print("\nFirst failure above supported limit:")
        print(first_failure_after_limit["benchmark_folder"])
        print("Contrast [%]:", first_failure_after_limit["folder_contrast_percent_relative_to_silicon"])
        print("Failure class:", first_failure_after_limit["quality_class"])
```

```python
# ============================================================
# CELL 9 — FINAL SCIENTIFIC LIMIT TABLE
# ============================================================

if len(passed_cases) > 0:
    final_conclusion = (
        f"The FWI method supports circular void reconstruction up to approximately "
        f"{max_supported_contrast_percent:.1f}% speed contrast relative to silicon. "
        f"This corresponds to a maximum supported speed difference of "
        f"{max_supported_speed_difference:.2f} m/s from the silicon background speed. "
        f"Above this contrast, reconstruction quality degrades and fails the selected "
        f"amplitude, localization, artifact, wrong-sign, shape-overlap, and background-error criteria."
    )
else:
    final_conclusion = (
        "No tested contrast passed the selected quality criteria. "
        "The current FWI configuration did not demonstrate reliable circular-void reconstruction."
    )

limit_summary_df = pd.DataFrame([{
    "silicon_background_speed_m_per_s": si_speed,

    "maximum_supported_contrast_percent_relative_to_silicon": max_supported_contrast_percent,
    "maximum_supported_speed_difference_m_per_s": max_supported_speed_difference,

    "final_scientific_conclusion": final_conclusion,

    "quality_criteria_recovery_min_percent": recovery_min_percent,
    "quality_criteria_recovery_max_percent": recovery_max_percent,
    "quality_criteria_min_detection_efficiency": min_detection_efficiency,
    "quality_criteria_max_false_positive_ratio": max_false_positive_ratio,
    "quality_criteria_max_wrong_sign_ratio": max_wrong_sign_ratio,
    "quality_criteria_min_IoU": min_iou,
    "quality_criteria_min_F1": min_f1_score,
    "quality_criteria_background_MAE_fraction": max_background_mae_fraction,
    "range_tolerance_m_per_s": range_tolerance_m_per_s,
}])

limit_csv = os.path.join(
    comparison_dir,
    "FINAL_maximum_supported_void_contrast.csv"
)

limit_summary_df.to_csv(limit_csv, index=False)

display(limit_summary_df)

print("Saved:")
print(limit_csv)
print()
print("FINAL SCIENTIFIC CONCLUSION:")
print(final_conclusion)
```

```python
# ============================================================
# CELL 9B — CONTRAST-BY-CONTRAST CONCLUSION TABLE
# ============================================================

contrast_conclusion_rows = []

for _, row in stage2_summary.iterrows():

    contrast = row["folder_contrast_percent_relative_to_silicon"]
    void_speed = row["folder_true_void_speed_m_per_s"]
    speed_difference = row["folder_speed_difference_from_silicon_m_per_s"]

    if row["overall_quality_pass"]:
        conclusion = "acceptable_reconstruction"
    elif contrast > max_supported_contrast_percent:
        conclusion = "beyond_supported_limit"
    else:
        conclusion = "failed_within_tested_range"

    contrast_conclusion_rows.append({
        "contrast_percent_relative_to_silicon": contrast,
        "void_speed_m_per_s": void_speed,
        "speed_difference_from_silicon_m_per_s": speed_difference,

        "void_recovery_percent": row["void_velocity_update_recovery_percent"],
        "void_MAE_m_per_s": row["void_MAE_m_per_s"],
        "background_MAE_m_per_s": row["background_MAE_m_per_s"],

        "IoU": row["IoU_intersection_over_union"],
        "F1_score": row["F1_score"],

        "wrong_sign_ratio_inside_void": row["wrong_sign_ratio_inside_void"],
        "false_positive_ratio": row["false_positive_ratio_background_to_void"],

        "overall_quality_pass": row["overall_quality_pass"],
        "quality_class": row["quality_class"],
        "scientific_conclusion": conclusion,
    })

contrast_conclusion_df = pd.DataFrame(contrast_conclusion_rows)

contrast_conclusion_csv = os.path.join(
    comparison_dir,
    "FINAL_contrast_by_contrast_conclusions.csv"
)

contrast_conclusion_df.to_csv(contrast_conclusion_csv, index=False)

display(contrast_conclusion_df)

print("Saved:")
print(contrast_conclusion_csv)
```

```python
# ============================================================
# CELL 12 — SHAPE LOCALIZATION VS CONTRAST
# ============================================================

x = stage2_summary["folder_contrast_percent_relative_to_silicon"]

plt.figure(figsize=(8, 5))

plt.plot(
    x,
    stage2_summary["IoU_intersection_over_union"],
    marker="o",
    label="IoU"
)

plt.plot(
    x,
    stage2_summary["F1_score"],
    marker="o",
    label="F1 score"
)

plt.axhline(min_iou, linewidth=1, linestyle="--", label="IoU threshold")
plt.axhline(min_f1_score, linewidth=1, linestyle=":", label="F1 threshold")

if not np.isnan(max_supported_contrast_percent):
    plt.axvline(
        max_supported_contrast_percent,
        linewidth=2,
        linestyle="--",
        label="Max supported contrast"
    )

plt.xlabel("Void speed contrast relative to silicon [%]")
plt.ylabel("Shape score [0 to 1]")
plt.title("Void localization quality versus contrast")
plt.ylim(0, 1.05)
plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
save_current_figure("03_shape_quality_vs_contrast")
plt.show()
```

```python
# ============================================================
# CELL 14 — PASS / FAIL DECISION VS CONTRAST
# 
# ============================================================

x = stage2_summary["folder_contrast_percent_relative_to_silicon"]
y = stage2_summary["overall_quality_pass"].astype(int)

plt.figure(figsize=(9, 5))

plt.scatter(
    x,
    y,
    s=90
)

for _, row in stage2_summary.iterrows():

    x_pos = row["folder_contrast_percent_relative_to_silicon"]
    y_pos = int(row["overall_quality_pass"])
    label = row["quality_class"]

    if row["overall_quality_pass"]:
        # PASS labels below the point
        plt.text(
            x_pos,
            y_pos - 0.08,
            label,
            rotation=0,
            ha="center",
            va="top",
            fontsize=9
        )
    else:
        # failure labels above the point
        plt.text(
            x_pos,
            y_pos + 0.08,
            label,
            rotation=45,
            ha="left",
            va="bottom",
            fontsize=8
        )

if not np.isnan(max_supported_contrast_percent):
    plt.axvline(
        max_supported_contrast_percent,
        linewidth=2,
        linestyle="--",
        label="Max supported contrast"
    )

plt.yticks([0, 1], ["Fail", "Pass"])
plt.ylim(-0.25, 1.25)

plt.xlabel("Void speed contrast relative to silicon [%]")
plt.ylabel("FWI quality decision")
plt.title("Practical FWI contrast limit: reliable up to about 50%")
plt.grid(True, alpha=0.3)
plt.legend(loc="upper right")

plt.tight_layout()
save_current_figure("05_pass_fail_vs_contrast_fixed")
plt.show()
```

```python
# ============================================================
# CELL 16 — TRUE VS RECONSTRUCTED VOID SPEED BY CONTRAST
# ============================================================

plot_df = stage2_summary.sort_values(
    "folder_contrast_percent_relative_to_silicon"
).copy()

x = plot_df["folder_contrast_percent_relative_to_silicon"]

plt.figure(figsize=(9, 5))

plt.plot(
    x,
    plot_df["folder_true_void_speed_m_per_s"],
    marker="o",
    label="True void speed"
)

plt.plot(
    x,
    plot_df["reconstructed_mean_void_speed_m_per_s"],
    marker="o",
    label="Stage 2 reconstructed mean void speed"
)

plt.axhline(
    si_speed,
    linewidth=2,
    linestyle=":",
    label="Silicon background speed"
)

plt.xlabel("Void speed contrast relative to silicon [%]")
plt.ylabel("Sound speed [m/s]")
plt.title("True vs reconstructed void speed by contrast")
plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
save_current_figure("06_true_vs_reconstructed_void_speed_by_contrast")
plt.show()
```
